```python
x = torch.tensor([2.0, 3.0], requires_grad=True)
print(x)

y = x + 2
print(y)

z = y * y * 3
z = z.mean()
print(z)

z.backward() # dz/dx
print("Gradients of z with respect to x: ", x.grad)
```
